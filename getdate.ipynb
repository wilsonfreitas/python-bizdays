# getdate

`Calendar.getdate(expr, ref)` resolves date expressions against month, year, or date references. It returns `numpy.datetime64` for scalar calls and `numpy.ndarray` for vectorized calls.

The `expr` argument has three forms:

- `<position> <day>`, such as `first day`, `15th day`, `last bizday`, `next wed`
- `<position> <day> <before|after> <position> <day>`, such as `first day before 15th day` or `1st bizday before 2nd fri`
- `previous <weekday>`, such as `previous mon`

Positions are `next`, `first`/`1st`, `second`/`2nd`, `third`/`3rd`, `<n>` with the right English suffix (`4th`, `11th`, `21st`, `22nd`) and `last`. Days are `day`, `bizday`, or a weekday as a 3-letter or full English name (`mon`/`monday` ... `sun`/`sunday`). Words are case-insensitive; any other word raises `ValueError`.

Reference rules:

- year or month refs: the n-th day, business day or weekday inside the period; `last` is the last one and `next` means the first. Composite expressions work with period refs only.
- date refs: `<n> day` is the date plus n days, `<n> bizday` is `offset(date, n)`, and `<n> <weekday>` is the n-th such weekday strictly after the date. `last` (or `previous`, for weekdays) gives the one strictly before.
- all refs in one call must be of one kind (year, month or date).

```python
from bizdays import Calendar

cal = Calendar.load(name="ANBIMA")
```

Month references use `YYYY-MM` strings.

```python
cal.getdate("15th day", "2002-05")
```

```python
cal.getdate("first day before 15th day", "2002-01")
```

```python
cal.getdate(["last day", "last bizday"], "2006")
```

The last business day of a month. All refs in one call must be of one kind: mixing month and date refs raises `ValueError`.

```python
cal.getdate("last bizday", "2024-02")
```

Year references can be integers or year strings.

```python
cal.getdate("first bizday", 2002)
```

Date references resolve the expression relative to that day.

```python
cal.getdate("next wed", "2021-02-10")
```

```python
cal.getdate("previous mon", "2021-02-10")
```

```python
cal.getdate("second fri", "2026-04-20")
```

```python
cal.getdate(["next bizday", "last bizday"], "2022-06-18")
```

```python
cal.getdate("last mon", "2022-06-20")
```

Both the expression and the reference can be vectorized, with refs of one kind per call.

```python
cal.getdate(["15th day", "16th day"], ["2002-01", "2001-01"])
```

```python
cal.getdate(["next wed", "previous mon"], ["2021-02-10", "2021-02-10"])
```
