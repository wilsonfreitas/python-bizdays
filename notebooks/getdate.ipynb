# getdate

`Calendar.getdate(expr, ref)` resolves date expressions against month, year, or date references. It returns `numpy.datetime64` for scalar calls and `numpy.ndarray` for vectorized calls.

The `expr` argument follows the parser implemented in `bizdays/getdate_parser.py`. It accepts three expression families.

Simple expressions use:

- `<ordinal> <target>`
- examples: `first day`, `15th day`, `last bizday`, `third tue`

Composite expressions use:

- `<ordinal> <target> <before|after> <ordinal> <target>`
- examples: `first day before 15th day`, `second bizday after 15th day`, `1st bizday before 2nd fri`

Date-relative expressions use:

- `<next|previous> <weekday>`
- examples: `next wed`, `previous mon`

Valid `ordinal` tokens are named ordinals such as `first`, `second`, `third`, `last`, plus numeric ordinals such as `1st`, `2nd`, `3rd`, `4th`, `15th`, and so on.

Valid `target` tokens are:

- `day`
- `bizday`
- weekday names recognized by their first three letters: `mon`, `tue`, `wed`, `thu`, `fri`, `sat`, `sun`

Reference rules:

- month refs (`YYYY-MM`) support simple and composite expressions
- year refs (`YYYY` or `int`) support simple and composite expressions
- date refs (`YYYY-MM-DD` and date-like values) support weekday-based relative expressions such as `next wed` and `previous mon`

`next` and `previous` require a weekday target, and composite expressions only allow the operators `before` and `after`.

```python
from bizdays import Calendar

cal = Calendar.load(name="ANBIMA")
```

Month references use `YYYY-MM` strings.

```python
cal.getdate("15th day", "2002-05")
```

```python
cal.getdate("first day before 15th day", "2002-01")
```

```python
cal.getdate(["last day", "last bizday"], "2006")
```

Year references can be integers or year strings.

```python
cal.getdate("first bizday", 2002)
```

Date references are useful for relative weekday expressions.

```python
cal.getdate("next wed", "2021-02-10")
```

```python
cal.getdate("previous mon", "2021-02-10")
```

```python
cal.getdate("second fri", "2026-04-20")
```

Both the expression and the reference can be vectorized.

```python
cal.getdate(["first day", "next wed"], ["2002-01", "2021-02-10"])
```
