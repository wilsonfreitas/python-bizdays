# python-bizdays overview

This notebook mirrors the current public documentation story: packaged calendars, constructor-based calendars, JSON calendar files, and the NumPy-backed `Calendar` API.

```python
from bizdays import Calendar

cal = Calendar.load(name="ANBIMA")
cal.bizdays("2014-01-13", "2015-01-13")
```

```python
custom = Calendar(
    holidays=["2024-01-01", "2024-12-25"],
    weekdays=["Saturday", "Sunday"],
    name="Example",
)
custom.following("2024-12-25")
```

Use `Calendar.load(name="...")` for packaged calendars, `Calendar.load(filename="...")` for custom JSON files, and the notebooks under `docs/source/` for the full method-by-method reference and migration guide.
