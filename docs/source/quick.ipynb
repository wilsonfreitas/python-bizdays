# Quickstart

The top-level `bizdays.Calendar` is the current public API. A single calendar instance can work with scalar inputs, vectorized inputs, NumPy arrays, and pandas date-like values.

```python
from bizdays import Calendar

cal = Calendar.load(name="ANBIMA")
cal
```

Business-day counts are available through `bizdays`.

```python
cal.bizdays("2014-01-13", "2015-01-13")
```

Date adjustment methods return NumPy dates.

```python
cal.following("2015-12-25")
```

Vectorized methods return NumPy arrays.

```python
cal.offset(["2014-01-02", "2014-01-03"], [1, 5])
```

```python
cal.seq("2014-01-02", "2014-01-07")
```

`getdate` resolves date expressions against month, year, or date references.

```python
cal.getdate("15th day", "2002-05")
```
