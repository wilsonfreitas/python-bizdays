# Quickstart

This page shows bizdays 2.0. A single `bizdays.Calendar` instance works with scalar inputs, vectorized inputs, NumPy arrays, and pandas date-like values.

```python
from bizdays import Calendar

cal = Calendar.load(name="ANBIMA")
cal
```

Business-day counts are available through `bizdays`.

```python
cal.bizdays("2014-01-13", "2015-01-13")
```

Date adjustment methods return NumPy dates.

```python
cal.following("2015-12-25")
```

Vectorized methods return NumPy arrays.

```python
cal.offset(["2014-01-02", "2014-01-03"], [1, 5])
```

```python
cal.seq("2014-01-02", "2014-01-07")
```

`getdate` resolves date expressions against month, year, or date references.

```python
cal.getdate("15th day", "2002-05")
```

`getbizdays` counts the business days in a whole year, or in one month.

```python
cal.getbizdays(2024)
```

Missing values give missing results. A vector of flags is a masked array: a NumPy array with a mask that marks the missing elements (shown as `--`).

```python
cal.isbizday(["2024-01-01", None, "2024-01-02"])
```
